```python
import time
import requests

url = "https://noobylub--qwen35-9b-lora-serving-server.us-east.modal.direct".rstrip("/")
deadline = time.monotonic() + 1200
while True:
    response = requests.get(f"{url}/v1/models", timeout=30)
    if response.status_code != 503:
        response.raise_for_status()
        print(response.json())
        break
    if time.monotonic() >= deadline:
        raise TimeoutError("Server did not start; check Modal logs.")
    time.sleep(5)
```
